During first initial approach to a new dataset, examining a feature correlation heatmap is like looking at a radar scan of how the variables interact with each other.

In machine learning, we don't just care about how features relate to the target variable (churn_value); we care deeply about how features relate to each other. Looking at the customer churn heatmap, there are four major reasons why this is an essential first step before training any models.

1. Spotting Multicollinearity (The "Double Counting" Problem)

    Multicollinearity happens when two or more independent variables provide nearly identical information.

    * Look at the heatmap: Notice the bright peach/orange squares where months (tenure), contract, and monthly intersect. Similarly, see how security, backup, protection, and support all share a warm orange correlation with each other.

    * Why it matters: If two features are highly correlated, individual models like Logistic Regression or Decision Trees get confused. For Logistic Regression, it makes the coefficient weights unstable and mathematically unreliable. For a standard Decision Tree, it splits redundantly.

    * The Fix: If two features have a correlation close to 1.0 or −1.0, you often drop one of them or combine them to prevent "double counting" information.

2. Identifying Strong Predictors for the Target

    A heatmap lets you immediately see which levers actually drive your target variable (churn_value).

    * Look at your map: Find churn_value on the y-axis and trace it horizontally.

        * It has a strong negative (very dark/black) correlation with satisfaction. This makes perfect sense: as customer satisfaction goes down, churning goes up.

        * It has a positive (light orange) correlation with monthly charges and internet_type_Fiber Optic.

    * Why it matters: This gives you an immediate baseline expectation. You now know that satisfaction, monthly, and internet_type are likely going to be the most important features when you build your models later.

3. Catching Redundant Dummy Variables (One-Hot Encoding Artifacts)

    When you convert categorical data into numbers using One-Hot Encoding, you can accidentally introduce perfect negative correlations.

    * Look at your map: Look at the intersection of internet_type_None and monthly or unlimited. They have a near-perfect dark black negative correlation (around −0.7).

    * Why it matters: If a customer has no internet type, their data for "unlimited data plan" or "monthly internet charges" will inherently drop to zero. This visual proof reminds you that these features are mechanically tied together, which helps you understand why a model might treat them as redundant.

4. Guiding Your Feature Engineering Strategy

    The heatmap helps you cluster behaviors. For instance, the fact that security, backup, protection, and support are all moderately correlated with each other suggests that customers usually buy tech services in bundles.

    * Why it matters: Instead of forcing your model to look at four separate columns, this correlation trend tells you that you could engineer a new, single feature called total_add_on_services_count by summing them up. This simplifies your data and often boosts model performance.

<BR>

#### How This Connects to the Classifiers

Depending on which model you choose next, this heatmap tells you exactly what hurdles you will face:

* Logistic Regression: Will struggle significantly with the correlated tech support features unless you remove the redundancy or use regularization (like Lasso/Ridge) to penalize them.

* Decision Trees: Will easily handle these correlations, but the tree might become unnecessarily deep and overfit by splitting on redundant features sequentially.

* Random Forest: Handles this beautifully because its feature-sampling randomness forces individual trees to look at security in one tree and backup in another, effectively balancing out the correlation!




### MultiCollinearity and Random Forest Algorithm

In this tutorial, we are focusing on Random Forest algorithm. This specific model we are using, Random Forest, is mathematically immune to the dangers of multicollinearity. Thus for the rest of the tutorial we will not be looking at the features correlations any further. There is no need to drop any correlated features because Random Forest handles them automatically.

<BR>

#### Summary Guide: When to use the Heatmap on Features Correlations

|Classifier Family|Do you need to drop correlated features using the Heatmap?|Why?|
|---|---|---|
|"Linear Models (Logistic Regression, Linear SVM)"|"Yes, absolutely."|Prevents unstable weights and mathematical confounding.|
|"Distance Models (KNN, RBF-Kernel SVM)"|"Yes, highly recommended."|Prevents the correlated features from hijacking the distance math.|
|Single Trees (Decision Tree)|Optional.|"Won't ruin accuracy, but dropping them makes the tree cleaner to read."|
|"Tree Ensembles (Random Forest, Extra Trees)"|"No, completely optional."|Internal feature random-subsetting naturally neutralizes the issue.|

<h2 id="TargPreproc">Part 2: Examining the Target and Preprocessing</h2>

In this exercise, we will proceed as follows:

*   Examine distribution of the predicted variable (`churn_value`).
*   Split the data into train and test sets. Decide if a stratified split should be used or not based on the distribution.
*   Examine the distribution of the predictor variable in the train and test data.


```python
# Data are skewed at ~85% towards non-churned customers
# This will be important to remember when model building
target = 'churn_value'
data[target].value_counts()
```

```python
data[target].value_counts(normalize=True)
```

Given the skew in the predictor variable, let's split the data with the *churned* values being stratified to ensure that the test and train sets have similar distributions.


```python
from sklearn.model_selection import StratifiedShuffleSplit


feature_cols = [x for x in data.columns if x != target]


# Split the data into two parts with 1500 points in the test data
# This creates a generator
strat_shuff_split = StratifiedShuffleSplit(n_splits=1, test_size=1500, random_state=42)

# Get the index values from the generator
train_idx, test_idx = next(strat_shuff_split.split(data[feature_cols], data[target]))

# Create the data sets
X_train = data.loc[train_idx, feature_cols]
y_train = data.loc[train_idx, target]

X_test = data.loc[test_idx, feature_cols]
y_test = data.loc[test_idx, target]
```

Let's check the distributions:


```python
y_train.value_counts(normalize=True)
```

```python
y_test.value_counts(normalize=True)
```

<h2 id ="OOB">Part 3: Random Forest and Out-of-bag(oob) Error</h2>

In this exercise, we will:

*   Fit random forest models with a range of tree numbers and evaluate the out-of-bag error for each of these models.
*   Plot the resulting oob errors as a function of the number of trees.

*Note:* since the only thing changing is the number of trees, the `warm_start` flag can be used so that the model just adds more trees to the existing model each time. Use the `set_params` method to update the number of trees.


```python
# Suppress warnings about too few trees from the early models
# import warnings
# warnings.filterwarnings("ignore", category=UserWarning)
# warnings.filterwarnings("ignore", category=RuntimeWarning)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize the random forest estimator
# Note that the number of trees is not setup here
RF = RandomForestClassifier(oob_score=True, 
                            random_state=42, 
                            warm_start=True,
                            n_jobs=-1)

oob_list = list()

# Iterate through all of the possibilities for number of trees
for n_trees in [30, 40, 50, 100, 150, 200, 300, 400]:    # will generate warning if less than 30; unless we suppress the warning previously
    RF.set_params(n_estimators=n_trees)  # Use this to set the number of trees
    RF.fit(X_train, y_train)  # Fit the model

    oob_error = 1 - RF.oob_score_   # Get the oob error
    oob_list.append(pd.Series({'n_trees': n_trees, 'oob': oob_error}))  # Store it

rf_oob_df = pd.concat(oob_list, axis=1).T.set_index('n_trees')
rf_oob_df
```

Looking at the above table, the Out-Of-Bag (OOB) error values are hovering around 0.046 to 0.049. In plain terms, our Random Forest model has an estimated error rate of roughly 4.7% to 4.9%, which means it is achieving an estimated accuracy of ~95.2% (1−error). Because OOB scores are calculated using data rows that each individual tree never saw during training, this acts as a built-in cross-validation. We can highly trust that these error rates reflect how the model will perform on completely unseen test data.

The error looks like it has stabilized around 100-150 trees.


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_context('talk')
sns.set_style('white')

ax = rf_oob_df.plot(legend=False, marker='o', figsize=(10, 6), linewidth=3)
ax.set(ylabel='out-of-bag error');
```

<h2 id = "ExtraTrees">Part 4: Extra Trees</h2>

Our exercise:

*   Repeat Part 3 above using extra randomized trees (`ExtraTreesClassifier`). Note that the `bootstrap` parameter will have to be set to `True` for this model.
*   Compare the out-of-bag errors for the two different types of models.


```python
from sklearn.ensemble import ExtraTreesClassifier

# Initialize the random forest estimator
# Note that the number of trees is not setup here
EF = ExtraTreesClassifier(oob_score=True, 
                          random_state=42, 
                          warm_start=True,
                          bootstrap=True,
                          n_jobs=-1)

oob_list = list()

# Iterate through all of the possibilities for 
# number of trees
for n_trees in [30, 40, 50, 100, 150, 200, 300, 400]:
    
    # Use this to set the number of trees
    EF.set_params(n_estimators=n_trees)
    EF.fit(X_train, y_train)

    # oob error
    oob_error = 1 - EF.oob_score_
    oob_list.append(pd.Series({'n_trees': n_trees, 'oob': oob_error}))

et_oob_df = pd.concat(oob_list, axis=1).T.set_index('n_trees')

et_oob_df
```

Combine the two dataframes into a single one for easier plotting.


```python
oob_df = pd.concat([rf_oob_df.rename(columns={'oob':'RandomForest'}),
                    et_oob_df.rename(columns={'oob':'ExtraTrees'})], axis=1)

oob_df
```

The random forest model performs consistently better than the extra randomized trees.


```python
sns.set_context('talk')
sns.set_style('white')

ax = oob_df.plot(marker='o', figsize=(10, 6), linewidth=2)
ax.set(ylabel='out-of-bag error');
```

The above plots help to visualize the core theoretical difference between Random Forest and Extra Trees in action on our specific dataset.

Here is exactly how to read what is happening in this graph.
1. Random Forest consistently beats Extra Trees here

    The blue line (Random Forest) sits lower on the y-axis than the orange line (Extra Trees) across the entire range of tree counts.

    * Meaning: For this specific cell dataset, Random Forest has a lower out-of-bag error rate (~4.67% at its best) compared to Extra Trees (~4.85% at its best).

    * The Math Behind It: Remember that Extra Trees chooses split thresholds completely at random rather than calculating the mathematically optimal split point (like Gini or Entropy) among the selected features. This extreme randomness introduces higher bias into the individual trees. On your dataset, that extra randomness makes the model slightly less precise than a Random Forest.

2. Extra Trees needs more trees to stabilize

    Look at how steep the orange line is at the beginning (from 30 to 100 trees) compared to the blue line.

    * Meaning: Because Extra Trees splits randomly, individual trees make a lot of wild mistakes early on.

    * The Ensemble Fix: It requires a larger crowd of trees to average out those wild, random errors. Notice how the orange line drastically drops and finally gets close to the blue line once you hit 150 to 200 trees.

3. The "Sweet Spot" is 150 Trees for both

    Both models reach their absolute lowest error rate at exactly 150 trees.

    * Past 150 trees, both curves start to bounce upward slightly or flatten out. This tells you that adding a 200th, 300th, or 400th tree isn't helping the ensemble anymore; the variance has already been minimized as much as it can be.


### Which model should you choose?

Based purely on this graph, Random Forest with 150 trees is our clear winner for performance, as it yields the lowest overall error rate (~4.67%).

* The One Saving Grace for Extra Trees: Even though Extra Trees loses slightly on accuracy here, it is computationally much faster to train on massive datasets. Because it doesn't spend CPU cycles calculating the absolute best mathematical split point for every node, it just picks a random number and moves on.

<h2 id="results">Part 5: Gathering Results</h2>

Here, we will:

*   Select one of the models that performs well and calculate error metrics and a confusion matrix on the test data set.
*   Given the distribution of the predicted class, which metric is most important? Which could be deceiving?


```python
# Random forest with 150 estimators
model = RF.set_params(n_estimators=150)

y_pred = model.predict(X_test)
```

Unsurprisingly, recall is rather poor for the customers who churned (True) class since they are quite small. We are doing better than random guessing, though, as the accuracy is 0.94 (vs 0.85 for random guessing).


```python
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score
from sklearn.metrics import f1_score, roc_auc_score

cr = classification_report(y_test, y_pred)
print(cr)

score_df = pd.DataFrame({'accuracy': accuracy_score(y_test, y_pred),
                         'precision': precision_score(y_test, y_pred),
                         'recall': recall_score(y_test, y_pred),
                         'f1': f1_score(y_test, y_pred),
                         'auc': roc_auc_score(y_test, y_pred)},
                         index=pd.Index([0]))

print()
print(score_df)
```

<h2 id="examineRes">Part 6: Examining Results</h2>

The following exercises will help us examine results:

*   Print or visualize the confusion matrix.
*   Plot the ROC-AUC and precision-recall curves.
*   Plot the feature importances.


```python
from sklearn.metrics import roc_curve, precision_recall_curve, confusion_matrix,ConfusionMatrixDisplay

sns.set_context('talk')
cm = confusion_matrix(y_test, y_pred, normalize='true')
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
disp.plot()
plt.show()
```

The ROC-AUC and precision-recall curves.


```python
sns.set_context('talk')

fig, axList = plt.subplots(ncols=2)
fig.set_size_inches(12, 6)

# Get the probabilities for each of the two categories
y_prob = model.predict_proba(X_test)

# Plot the ROC-AUC curve
ax = axList[0]

fpr, tpr, thresholds = roc_curve(y_test, y_prob[:,1])
ax.plot(fpr, tpr, linewidth=3)
# It is customary to draw a diagonal dotted line in ROC plots.
# This is to indicate completely random prediction. Deviation from this
# dotted line towards the upper left corner signifies the power of the model.
ax.plot([0, 1], [0, 1], ls='--', color='black', lw=.3)
ax.set(xlabel='False Positive Rate',
       ylabel='True Positive Rate',
       xlim=[-.01, 1.01], ylim=[-.01, 1.01],
       title='ROC curve')
ax.grid(True)

# Plot the precision-recall curve
ax = axList[1]

precision, recall, _ = precision_recall_curve(y_test, y_prob[:,1])
ax.plot(recall, precision, linewidth=3)
ax.set(xlabel='Recall', ylabel='Precision',
       xlim=[-.01, 1.01], ylim=[-.01, 1.01],
       title='Precision-Recall curve')
ax.grid(True)

plt.tight_layout()
```

Now, let's look into the importance of features. Apparently, "satisfaction" is the most significant predictor of customer churn.


```python
feature_imp = pd.Series(model.feature_importances_, index=feature_cols).sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(16, 6))
ax.pie(feature_imp, labels=None, autopct=lambda pct: '{:1.1f}%'.format(pct) if pct > 5.5 else '')
ax.set(ylabel='Relative Importance')
ax.set(xlabel='Feature')

# Adjust the layout to prevent label overlapping
plt.tight_layout()

# Move the legend outside the chart
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5),labels=feature_imp.index)

plt.show()
```
