# Gradient Boosting  for classification with Python

Estimated time needed: **1.45** hours


In this notebook, you will learn Gradient Boosting for classification; AdaBoost is a particular case of Gradient Boosting and an additive model where we add weak learners to minimize the loss function. This lab will focus on <a href="https://xgboost.readthedocs.io/en/stable/?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMML241ENSkillsNetwork31576874-2022-01-01">XGBoost</a>, an open-source software library that provides a regularizing gradient boosting framework.  XGBoost can use different types of weak learners called boosters for classification and regression. We will focus on trees for classification.


Unlike Bagging and Random Forest, Gradient Boosting can cause overfitting. As a result, Gradient Boosting requires Hyperparameter tuning, taking more time to train. One advantage of Gradient Boosting is that each classifier is smaller, so predictions are faster in Gradient Boosting.
AdaBoost is a subclass of Gradient Boosting; one weakness of AdaBoost is that misclassified samples cause overfitting; Gradient Boosting uses different loss functions, reducing this effect.
The following table show's the Average accuracy and standard deviation for the Random Forest (RF), gradient boosting (GB) and XGBoost, both using the default (D) and (T) tuned parameter settings, we see that XGBoost does best, followed by GB outperforming the other methods.

### Comparing Gradient Boosting with Stacking

While Stacking acts like a panel of different experts whose opinions are blended by a final manager, Gradient Boosting is an assembly line of specialists built sequentially, where each new specialist is trained exclusively to fix the exact mistakes made by the ones before it.

Here is a simple way to visualize the core difference between the two approaches:

1. **How Gradient Boosting Works**

    Gradient Boosting builds a sequence of small, simple models—typically shallow decision trees called "weak learners"—one after another.


    [Base Model 1] ──> Calculates Errors (Residuals)

                            │
                            ▼
                    [Base Model 2] ──> Learns to predict Model 1's errors

                                            │
                                            ▼
                                    [Base Model 3] ──> Learns to predict Model 2's errors

* Step 1: Train a very basic decision tree on your target variable ($y$). It makes a reasonable initial guess, but makes some mistakes.

* Step 2: Calculate the errors (known as residuals) for every single prediction ($y - \hat{y}$).

* Step 3: Train a second decision tree, but not to predict $y$—its sole job is to predict those residuals (errors).

* Step 4: Add the predictions of the second tree to the first tree (scaled by a learning rate, e.g., $0.1$) to adjust for the mistakes.

* Repeat: Keep adding new trees, each one focusing entirely on the remaining leftover errors of the entire ensemble.

    The "Gradient" part comes from calculus: mathematically, predicting the residual errors is equivalent to taking steps down a Loss Function using gradient descent to reach optimal accuracy.

<BR>

2. **Gradient Boosting vs. Stacking: The Direct Comparison**

<BR>

|Dimension|Stacking|Gradient Boosting|
|---|---|---|
|Model Diversity|"Heterogeneous: Combines completely different algorithm types (e.g., SVM + KNN + Decision Tree)."|Homogeneous: Uses dozens or hundreds of the same base model (almost always small Decision Trees).|
|Execution Style|Parallel / Independent: Base models are trained independently on the raw dataset before the meta-model combines them.|Sequential / Dependent: Trees are built one after another. Tree #10 cannot be built until Trees #1–9 are finished.|
|Target Variable|Each base learner tries to predict the real target (y).|"Base Learner #1 predicts target (y). Learners #2, #3, #4... predict the error/residuals of the ensemble."|
|Combination Method|"A Meta-Model (e.g., Logistic Regression) learns weights to combine base outputs."|A weighted sum of all trees scaled by a learning rate (η).|

<BR>


3. **Popular Implementations**

    You would likely heard of the big three "flavor" variations of Gradient Boosting used in competitive machine learning:

    * XGBoost (Extreme Gradient Boosting): Known for extreme execution speed, built-in regularization, and handling sparse matrices smoothly.

    * LightGBM: Microsoft's framework that grows trees "leaf-wise" rather than "depth-wise," making it extremely fast on massive datasets.

    * CatBoost: Optimized specifically for handling categorical features without requiring manual LabelEncoder or OneHotEncoder preprocessing steps.

### How Learning Rate and the Number of Trees (n_estimators) control the bias-variance in XGBoost

#### The above 3-Run Breakdown

Run 1 (lr=0.1, n=5):    Train: 77.1%  |  Test: 76.7%  --> Underfitting<BR>
Run 2 (lr=0.3, n=5):    Train: 91.4%  |  Test: 78.3%  --> Sweet Spot (Best Test Score)<BR>
Run 3 (lr=0.3, n=100):  Train: 100%   |  Test: 70.0%  --> Severe Overfitting<BR>

<BR>

**Run 1: Underfitting (learning_rate=0.1, n_estimators=5)**

* What happened: We capped the ensemble at just 5 trees with a tiny step size (0.1).

* The result: The model was given way too little budget to learn the actual patterns in the data. Train accuracy (77.1%) and test accuracy (76.7%) sit right next to each other because the model hasn't learned enough complexity yet—it is suffering from high bias.

**Run 2: The Sweet Spot (learning_rate=0.3, n_estimators=5)**

* What happened: We kept the trees capped at 5, but tripled the step size (0.3). This forced each tree to make bolder corrections to the residuals.

* The result: Test accuracy peaked at 78.3%! The gap between train (91.4%) and test (78.3%) shows the model had enough capacity to capture genuine trends in the churn data without completely memorizing noise.

**Run 3: Severe Overfitting (learning_rate=0.3, n_estimators=100)**

* What happened: We jumped from 5 trees straight to 100 trees while keeping a high learning rate (0.3).

* The result: Train accuracy locked at 100%, but test accuracy collapsed to 70.0% (an 8.3% drop from Run 2).

* Why it failed: Unlike Random Forests (where adding more trees is safe because they are independent), XGBoost trees are additive. Adding 100 trees with a high learning rate meant the last 80+ trees spent their time memorizing the tiny, noisy edge cases of the 140 training rows.

<BR>

#### **3 Key Rules for Tuning XGBoost**

When tuning XGBoost in the future, keep this relationship in mind:

1. learning_rate and n_estimators move in opposite directions: If we decrease the learning rate (e.g., down to 0.01 or 0.05), we must increase n_estimators so the model has enough iterations to reach the optimal solution.

2. Use Early Stopping: Instead of guessing n_estimators, set it high (like 500) and use early_stopping_rounds=10. This automatically halts training the moment test/validation loss stops improving.

3. Small datasets need regularization: On small datasets like ChurnData (~200 rows), high tree counts will overfit rapidly unless you add tree-level constraints like max_depth=3 or reg_lambda=1.0 (L2 regularization).

In XGBoost, passing an eval_set (evaluation set) serves two vital purposes: monitoring real-time performance and preventing overfitting. As XGBoost sequentially builds tree after tree, it uses eval_set to test the growing ensemble on that validation data after every single tree is added.

We can input a  `eval_metric`  parameter  in the `fit` method; this  determines how the model performs on each iteration using a second dataset; in XG Boost the metrics include:


`logloss` : binary logarithmic loss

`mlogloss` cross entropy loss

`error` : number of misclassified samples

`auc` : area under  ROC curve (receiver operating characteristic curve)

As shown in above plot, here is what early stopping just achieved for your model:

What Changed?

1. Stopped Right at the Horizon: Instead of running blindly all the way to 100 trees and climbing up to $0.68$ loss, XGBoost calculated that validation loss hit its floor around tree 6 ($\text{loss} \approx 0.446$).

2. Patience Window: Because we set early_stopping_rounds=10, XGBoost gave the model 10 more "patience" rounds (from tree 6 to tree 16) to see if validation loss would drop again. When test loss kept creeping up toward $0.50$, it slammed the brakes at iteration 16.

When we call model.predict() now, XGBoost will automatically use the weights from tree 6 (its best iteration), completely protecting it from the overfitting that happened on trees 7 through 16!


We can lower the test loss curve even further (below 0.44) by dropping  the learning_rate from 0.3 down to 0.05 or 0.01. This is because a smaller learning rate takes tiny, cautious steps, the test loss curve will descend much more gracefully and usually finds a noticeably deeper "valley" before it starts rising again!


```python
model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.03,
    eval_metric='logloss',
    early_stopping_rounds=10  # Stops training if test loss doesn't improve for 10 trees
)

# Fit using the validation set
model.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_test, y_test)],
    verbose=True
)

print(f"Best iteration (optimal number of trees): {model.best_iteration}")

results = model.evals_result()

fig, ax = plt.subplots()
ax.plot( results['validation_0']['logloss'], label='Train')
ax.plot( results['validation_1']['logloss'], label='Test/Validation')
ax.legend()
plt.show()
```

Notice how smooth, gradual, and controlled that descent is compared to the 0.3 learning rate. By dropping the step size to 0.03, you gave the algorithm room to breathe and fine-tune its steps.

**Key Takeaways from This Plot:**

1. Gentle Descent: Instead of shooting down and immediately rebounding in under 10 trees, the test/validation loss gently descends and stabilizes across ~60–75 trees.

2. Stable Plateaus: The validation loss flattens out around 0.46 and stays virtually flat rather than aggressively spiking back up.

3. Smooth Generalization Gap: The distance between the train (blue) and test (orange) curves expands smoothly without dramatic erratic swings, meaning the model is learning generalized patterns far more reliably.

## How  Gradient Boosting  Works (Optional)


Let's try to sketch   how Gradient Boosting works like any supervised problem we have a dataset  ${(x_1, y_1), ..., (x_N,; y_N)} $, the strong classifier $H_{T}(x)$  is a linear combination of $T$  weak classifiers $h_t(x)$  usually trees and $\alpha_t$  is a constant in many cases $\alpha_t=1$  . Although each classifier $h_t(x)$ appears independent, the  contains information about the error of classifiers from $h_1(x),.., h_{t}(x)$.

$H_{T}(\mathbf{x}) =   \sum_{t=1}^T \alpha_t h_t(\mathbf{x}) $

Borrowing the notation form <a href=https://www.cs.cornell.edu/courses/cs4780/2018fa/lectures/lecturenote19.html?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMML241ENSkillsNetwork31576874-2022-01-01> Kilian Weinberger </a>, we have a the cost function:

$\ell(H)=\frac{1}{n}\sum_{i=1}^n \ell(H_{T}(\mathbf{x}_i),y_i)$

Where $\ell$ convex i.e bole shape shown below and differentiable, such as derivative exists.


```python
x = np.linspace(-1, 2)
fig, (ax1, ax2) = plt.subplots(1, 2)
ax1.plot(x, x**2, linewidth=2)
ax1.set_title('convex function')
ax2.plot(x, x**2 + np.exp(-5*(x - .5)**2), linewidth=2)
ax2.set_title('non-convex function')
plt.show()
```

If you're familiar with cost functions, this may seem strange as we are dealing with a function of $H$ not of parameters. Ideally would like to minimize the cost for  $h_{T}(\mathbf{x})$  and  $H_{T-1}(\mathbf{x})$ at the same time   , but this is difficult. So we minimize the cost with respect to   $h_{T}(\mathbf{x})$  while keeping $H_{T-1}(\mathbf{x})$ fixed; not only is this simpler it prevents overfitting.

$h_{t}(x) = 	extrm{argmin}_{   h \in \mathbb{H}} \sum_{i=1}^n \ell(  H_{t-1}(\mathbf{x}_i) + \gamma h_t(\mathbf{x}_i),y_i)$  (1)


Usually $\gamma$  is selected using validation data, but in some cases, like AdaBoost, you can find  the expression for the  optimal value of $\gamma$.


### How to Minimize Cost


We would like to find the value of classifier $h_{t}(x)$ in red  that minimises the Cost function in (1), this is difficult so we minimize the first order Taylor Approximation as shown in (2). We see that it is only a function of the second term in red.


$$\ell(H+\alpha h)\approx\ell(H)+ \gamma \sum_{i = 1}^{n} r_{i,t-1} h_{t}(\mathbf{x}_i) \quad (2)$$

$$r_{i,t-1}=\frac{\partial \ell}{\partial [H_{t-1}(\mathbf{x}_i)]}$$

An example of Taylor Approximation is shown in the following figure, the true cost function is in blue and the first Taylor Approximation is in Orange.


```python
plt.plot(x, x**2, linewidth=2)
plt.text(-.7, 3, r'$\ell(H_{T}+h_{t},y_i)$', size=20)
plt.plot(x, 2*x - 1)
plt.plot(1, 1, 'k+')
plt.text(.3, -.75, 'approximation', size=15)
plt.show()
```

For small values of $\gamma$ this will hold, using the approximation, we can decouple the problem as follows; the $argmax$ of $h$  only cares about the red terms. As a result, we find the value of $r_{i}$ and the minimum concerning $h$. First let's find $r_{i}$, we need the loss function:


$$h_{t}(\mathbf{x}) = \operatorname*{argmin}_{h \in \mathbb{H}} \sum_{i=1}^n \ell(H_{t-1}(\mathbf{x}_i) + \gamma h(\mathbf{x}_i), y_i)$$

$$\approx \operatorname*{argmin}_{h \in \mathbb{H}} \sum_{i = 1}^{n} r_{i} h(\mathbf{x}_i)$$

$$= \operatorname*{argmin}_{h \in \mathbb{H}} \left\{ \sum_{i = 1}^{n} r_{i} h(\mathbf{x}_i) \right\} \quad (3)$$

### Example with Python


This all seems a little confusing so let’s do a simple example with the Root mean square error loss are write the Python code. This is not usually used for regression but can be used for classification and  the math is relatively  simple to understand. First we calculate $r_{i}$:


$ \ell(H) =\sum_{i=1}^n (r_{i,t-1}-  H_{t-1}(\mathbf{x}_i))^{2} $

$ r_{i,t}=\frac{\partial \ell}{\partial [H_{T-1}(\mathbf{x}_i)]}=2\sum_{i=1}^n ( r_{i,t-1}-  H_{T-1}(\mathbf{x}_i))$


Then we plug it into equation 3; with some math, we get the following expression below, just a note these steps are not as simple as they look, so check out  <a href=https://www.cs.cornell.edu/courses/cs4780/2018fa/lectures/lecturenote19.html?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMML241ENSkillsNetwork31576874-2022-01-01> Kilian Weinberger </a> for more details.


$h_{t}(x) =	extrm{argmin}_{h \in \mathbb{H}}{ \sum_{i = 1}^{n} r_{i,t} \color{red}{h_{t}(\mathbf{x}_i)}}$

$=	extrm{argmin}_{h \in \mathbb{H}}{ \sum_{i = 1}^{n} 2\sum_{i=1}^n (y_i-  H_{T-1}(\mathbf{x}_i)) \color{red}{h_{T}(\mathbf{x}_i)}}$

$=	extrm{argmin}_{h \in \mathbb{H}}{ \sum_{i = 1}^{n} (r_{i,t}^{2}- {h_{t}(\mathbf{x}_i)})^{2}}$


Let’s go through a few iterations; at the same time implement the algorithm with  Python. For the Python portion we will use the Toy data where the class of y is as following $y=0$ if $ 0<x<1$ else if  $y=1$, $1 \leq x< 2$ else $y=3$.


```python
X=np.linspace(0, 3,num=100)
 
y=np.zeros(X.shape)
y[X>1]=1
y[X>2]=2
plt.plot(X,y)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

For the first iteration we start off with $h_1(\mathbf{x})$ minimizing the original labels $y_{i}$ that equals $r_{i,1}$.


$h_{1}(x)=	extrm{argmin}_{h \in \mathbb{H}}{ \sum_{i = 1}^{n} (y_{i}- {h_{1}(\mathbf{x}_i)})^{2}}$


All this means is we apply any algorithm with the cost function in the same form to the data. One constraint is  $h$ has to be non-linear, in Python:


```python
reg = DecisionTreeRegressor(max_depth=1)
reg.fit(X.reshape(-1,1),y)
h_1=reg.predict(X.reshape(-1,1))
```

We can plot the prediction, we see it's not even a true classifier.


```python
plt.plot(X,y,label="y")
plt.plot(X,h_1,label="$h_{1}(x)$")
plt.legend()
plt.show()
```

For the second step  we set  $H_{1}(\mathbf{x})=h_{1}(\mathbf{x})$,then we calculate $r_{i,2}$.


$ r_{i,2}=( r_{i,1}-  H_{1}(\mathbf{x}_i))$


```python
gamma=1
r=y-gamma*reg.predict(X.reshape(-1,1))
```

We then find the second learner $h_{2}(\mathbf{x})$ by minimizing


$h_{2}(x)=	extrm{argmin}_{h \in \mathbb{H}}{ \sum_{i = 1}^{n} (r_{i,2}^{2}- {h_{2}(\mathbf{x}_i)})^{2}}$


We then update $H_{2} (\mathbf{x})= H_{1} (\mathbf{x})+ \gamma h_{2}(\mathbf{x})$


We repeat the process for $ t$ times:


$ r_{i,t}^2=( r_{i,t-1}-  H_{t-1}(\mathbf{x}_i))^{2}$


$h_{t}(x)=	extrm{argmin}_{h \in \mathbb{H}}{ \sum_{i = 1}^{n} (r_{i,t}^{2}- {h_{t}(\mathbf{x}_i)})^{2}}$


We then update $H_{t} (\mathbf{x})= H_{t-1} (\mathbf{x}) + \gamma h_{t}(\mathbf{x})$


We can perform the Operation with a loop in Python. First, we will write a function to make a prediction using an input of a list of predictors `weak_learners`.


```python
def predict(y,weak_learners,gamma):
    yhat=np.zeros(y.shape)
    for h in weak_learners:
        yhat+=h.predict(X.reshape(-1,1))
    return yhat

```

Then we train a new weak learner recursively setting gamma=1, and plot the results for each iteration.


```python
weak_learners =[]
gamma=1
r=y
weak_learners.append(reg)
for t_ in range(0,10):
    #train weak learner 
    reg=DecisionTreeRegressor(max_depth=1)
    reg.fit(X.reshape(-1,1),r)
    weak_learners.append(reg)
    
    #Calculate r_i,t for each iteration  
    r=r-gamma*reg.predict(X.reshape(-1,1))
    #plot function   
    plt.plot(X,y,label="y")
    plt.plot(X,predict (y,weak_learners,gamma),label="$H_{}(x)$".format(t_+1))
    plt.legend()
    plt.show()
```
